## 2. Variables

Variables let us store values and reuse them later.



```python
spike_count = 12
trial_duration = 2.5  # seconds

print("Spike count:", spike_count)
print("Trial duration:", trial_duration)
```

We can combine variables in calculations.



```python
firing_rate = spike_count / trial_duration
print("Firing rate (spikes/s):", firing_rate)
```

## 3. Basic Python data types

Python has several built-in types. Here are a few common ones.



```python
integer_value = 5
float_value = 3.14
text_value = "neuroscience"
boolean_value = True

print(type(integer_value))
print(type(float_value))
print(type(text_value))
print(type(boolean_value))
```

## 4. Lists

Lists are ordered collections of items.



```python
spike_counts = [5, 8, 3, 10, 6]
print(spike_counts)
```

Length of a list



```python
len(spike_counts)
```

Accessing elements by index (Python starts counting at 0)



```python
print(spike_counts[0])
print(spike_counts[2])
print(spike_counts[-1])
```

## 5. Slicing

Slicing extracts parts of a list.



```python
print(spike_counts[0:3])
print(spike_counts[2:])
print(spike_counts[:4])
```

## 6. Simple functions

Python includes many useful built-in functions.



```python
print("Minimum:", min(spike_counts))
print("Maximum:", max(spike_counts))
print("Sum:", sum(spike_counts))
print("Mean:", sum(spike_counts) / len(spike_counts))
```

## 7. Simple loops

Loops let us repeat an operation for each item in a collection.



```python
for count in spike_counts:
    print("Spike count:", count)
```

Example: convert spike counts to firing rates for 1-second trials



```python
firing_rates = []
for count in spike_counts:
    rate = count / 1.0
    firing_rates.append(rate)

print(firing_rates)
```

## 8. Conditionals

Conditionals allow code to behave differently depending on a condition.



```python
threshold = 6

for count in spike_counts:
    if count > threshold:
        print(count, "is above threshold")
    else:
        print(count, "is not above threshold")
```

## 9. Working with strings

Strings are useful for labels, filenames, and messages.



```python
region_name = "Primary visual cortex"
print(region_name)
print(region_name.lower())
print(region_name.upper())
```

## 11. First figure with Matplotlib

Let's make a simple plot from our spike counts.



```python
plt.figure(figsize=(6, 4))
plt.plot(spike_counts, marker="o", label='example data')
plt.xlabel("Neuron index")
plt.ylabel("Spike count")
plt.title("Spike counts across neurons")
plt.show()
```

## 12. A tiny scientific workflow

A typical workflow is:
1. load or define data
2. inspect it
3. compute something
4. visualize the result



```python
reaction_times = [0.55, 0.61, 0.49, 0.72, 0.58, 0.63]
mean_rt = sum(reaction_times) / len(reaction_times)
print("Mean reaction time:", mean_rt)
```

```python
plt.figure(figsize=(6, 4))
plt.hist(reaction_times, bins=5)
plt.xlabel("Reaction time (s)")
plt.ylabel("Count")
plt.title("Distribution of reaction times")
plt.show()
```
